```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/titanic (1).csv")#to read the file
print(df)
```

```python
df.describe()#generate descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset's distribution, excluding NaN (null) values.
```

```python
df.isnull().sum()#is used to find the number of missing (null) values in each column of your DataFrame.
#returns true if null and false is there is data
#.sum() return true as 1 and false as 0
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"]).mode()[0]
# is meant to fill the missing values in your Embarked column with the most common value (the mode)

# 1. Check and sum up the number of missing (null) values in each column of the DataFrame
print(df.isnull().sum())

# 2. Get and print the number of rows in the DataFrame before dropping duplicates
print("before:", df.shape[0])#0 for row values and 1 for coloumn values

# 3. Remove all duplicate rows from the DataFrame and save the cleaned version back to 'df'
df = df.drop_duplicates()

# 4. Get and print the number of rows after removing duplicates to see if any were deleted
print("after:", df.shape[0])
```

```python


df = df.rename(columns={
    "Passengerid": "PassengerID",
    "2urvived": "survived"
})



df
```

```python
# 1. Access the 'Sex' column, which originally contains numeric values (0 and 1)
# 2. Use .map() with a dictionary to replace 0 with 'MALE' and 1 with 'FEMALE'
# 3. Assign the mapped/updated values back to overwrite the 'Sex' column
df["Sex"] = df["Sex"].map({0:"MALE", 1:"FEMALE"})
```

```python
sns.histplot(data=df,x='Age',bins=20) #ONLY VISUALIZE AGE
plt.title("Age distribution of titanic passengers")
plt.xlabel("Number of passengers")
plt.show()
```

```python
sns.countplot(data=df,x="Sex",hue="survived")
plt.title("titanic survival by sex")
plt.xlabel("passenger class")
plt.ylabel("number of passengers")
plt.show()
```

```python
sns.histplot(data=df,x='PassengerID',bins=20) #ONLY VISUALIZE PASSENGERID DATA

plt.title("Age distribution of titanic passengers")

plt.xlabel("Number of passengers")
plt.show()
```
